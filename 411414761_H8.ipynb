```python
import random
s = random.randint(1, 50)
for day in range(1, 4):
  print(f"今天第{day}天,請輸入1-50數字 ", end="")
  guess = int(input())
  print(f"今天的幸運數字是{s}")
  if guess == s:
    print("恭喜你中獎了")
    break
  else:
    print("你沒中獎，明天再來\n")
    s = random.randint(1, 50)
else:
  print("很遺憾3天都沒中獎，下次再試試吧！")
```
