```python
def c_bmi(h,w):
    bmi=w/(h/100)**2
    return bmi
bmi1=c_bmi(170,60)
bmi2=c_bmi(165,55)
bmi3=c_bmi(180,80)
print(f"第一位bmi{bmi1:.2f}\n第二位bmi{bmi2:.2f}\n第三位bmi{bmi3:.2f}")
```
