```python
def f1(p,q):
  return p*q
def f2(p,q):
  total=f1(p,q)
  return total*0.9
p=int(input("單價:"))
q=int(input("數量:"))
print("原價:",f1(p,q))
print("折扣:",f2(p,q))
```
